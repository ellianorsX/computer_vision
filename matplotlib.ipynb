```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([1, 3, 5, 7])

plt.plot(x, y1, label="Line 1", color="red", marker="s")
plt.plot(x, y2, label="Line 2", color="blue", linestyle="--", marker="o")
plt.title("Multiple Lines Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.legend()
plt.show()
```

# **Scatter Plot with Numpy and Matplotlib**

```python
x = np.array(["For Parents", "For Businnes", "For Charity", "For Dream"])
y = np.array([6, 9, 9, 5])

plt.bar(x, y, color="orange")
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 6, 3, 7])

plt.scatter(x, y, color="purple", s=100)
plt.title("Scatter Plot Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

# Histogram with Numpy and Matplotlib

```python
data = np.array([3, 4, 5, 6, 8, 8, 8, 9, 10])

plt.hist(data, color="skyblue", edgecolor="black")
plt.title("Histogram Example")
plt.xlabel("Values")
plt.ylabel("Count")
plt.show()
```

# Subplots(Multiple Graphs Together)

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([3, 6, 9, 12])

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y1, color="red")
plt.title("Plot 1")

plt.subplot(1, 2, 2)
plt.plot(x, y2, color="blue", marker="o")
plt.title("Plot 2")

plt.tight_layout()
plt.show()
```
